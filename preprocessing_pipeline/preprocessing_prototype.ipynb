```python
import pandas as pd
import numpy as np
import matplotlib.pylab as plt
import seaborn as sns
import time

from glob import glob
from itertools import cycle

import librosa
import librosa.display
import IPython.display as ipd
```

```python
class0 = glob("./data/Class0/*/*.mp3")
class1 = glob("./data/Class1/*/*.mp3")
```

```python
#normalize recordings to 0-1 scale
y, sr = librosa.load(class0[0])
y_norm = librosa.util.normalize(y)
```

```python
#remove silence
y_trim, _ = librosa.effects.trim(y_norm, top_db=20)
intervals = librosa.effects.split(y_trim, top_db=20)
audio_list = np.concatenate([y_trim[list[0]:list[1]] for list in intervals])
print(audio_list)
```

```python
clip_length = 3
clip_length_samples = clip_length*sr
max_len = sr*60
#crop into 1 minute recording
y_cropped = audio_list[:max_len]

#cut into 3s clips
y_clips = librosa.util.frame(y_cropped,
                             frame_length=clip_length_samples,
                             hop_length=clip_length_samples).T
print(len(y_clips))
# print(y_clips.shape)
ipd.Audio(y_clips[0], rate=sr)
```

```python
#spectograms
mel_spectograms = []
for clip in y_clips:
    S = librosa.feature.melspectrogram(y = clip, sr = sr, n_mels = 128,)
    S_db_mel = librosa.amplitude_to_db(S, ref=np.max)
    mel_spectograms.append(S_db_mel)

    ipd.clear_output(wait=True)
    fig, ax = plt.subplots(figsize = (10,5))
    img = librosa.display.specshow(S_db_mel, x_axis='time', y_axis='log', ax=ax, cmap='inferno')
    fig.colorbar(img, ax=ax, format = f'%0.2f')
    ipd.display(fig)
    S_pcen = librosa.pcen(S_db_mel, sr=sr)

```
